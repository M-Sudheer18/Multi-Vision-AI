## Step 1

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import cv2
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
from tensorflow.keras.utils import image_dataset_from_directory
```

## Step 2

```python
IMG_SIZE   = ( 32, 32 )
BATCH_SIZE = 64
```

```python
train_dir = "/kaggle/input/datasets/sudheermuthyala8688/image-data/cifar10/cifar10/train"
test_dir  = "/kaggle/input/datasets/sudheermuthyala8688/image-data/cifar10/cifar10/test"
```

```python
classes = os.listdir(train_dir)
for class_name in classes:
    train_count = len(os.listdir(os.path.join(train_dir, class_name)))
    test_count  = len(os.listdir(os.path.join(test_dir, class_name)))

    print(f"{class_name:12} Train: {train_count:5} | Test: {test_count:5}")
```

```python
train_dataset = image_dataset_from_directory(
    train_dir,
    image_size = IMG_SIZE,
    batch_size = BATCH_SIZE,
    label_mode = "int",
    shuffle = True
)

test_dataset = image_dataset_from_directory(
    test_dir,
    image_size = IMG_SIZE,
    batch_size = BATCH_SIZE,
    label_mode = "int",
    shuffle = False
)
```

```python
print(train_dataset.class_names)
print(test_dataset.class_names)
```

```python
print("Classes:", train_dataset.class_names)

for images, labels in train_dataset.take(1):
    print("Image batch shape:", images.shape)
    print("Label batch shape:", labels.shape)
```

```python
class_name = train_dataset.class_names
class_name
```

## Step 3

```python
norm_layer = tf.keras.layers.Rescaling(1./255)

train_norm = train_dataset.map(
    lambda images, labels: (norm_layer(images), labels)
)
test_norm  = test_dataset.map(
    lambda images, labels: (norm_layer(images), labels)
)
```

```python
for images, labels in train_norm.take(1):
    print(f"Max: {tf.reduce_min(images).numpy()}")
    print(f"Min: {tf.reduce_max(images).numpy()}")
    print(f"Shape: {images.shape}")
```
